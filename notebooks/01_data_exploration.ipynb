```python
import pandas as pd
import numpy as np

# Load data
df = pd.read_excel('../data/raw/ethiopia_fi_unified_data.xlsx')
ref = pd.read_excel('../data/raw/reference_codes.xlsx')

print(f"Dataset shape: {df.shape}")
df.head()
```

```python
# Understand the composition of the data
print(df['record_type'].value_counts())

# Check pillars
print("\nRecords by Pillar:")
print(df['pillar'].value_counts(dropna=False))
```

```python
# List unique indicators and their date ranges
indicator_summary = df[df['record_type'] == 'observation'].groupby('indicator_code').agg({
    'observation_date': ['min', 'max', 'count'],
    'value_numeric': ['mean', 'max']
})
print(indicator_summary)
```

```python
import datetime

# Example: Adding a new observation for Telebirr Users
new_observation = {
    'record_type': 'observation',
    'pillar': 'Usage',
    'indicator': 'Mobile Money Account Ownership',
    'indicator_code': 'ACC_MM_ACCOUNT',
    'value_numeric': 54.0,  # 54 Million
    'value_unit': 'Millions',
    'observation_date': '2024-05-01',
    'source_name': 'Ethio Telecom Press Release',
    'source_url': 'https://www.ethiotelecom.et/news/',
    'confidence': 'high',
    'collected_by': 'Your Name',
    'collection_date': datetime.date.today().isoformat(),
    'notes': 'Significant growth since 2021 launch'
}

# Append to dataframe
df = pd.concat([df, pd.DataFrame([new_observation])], ignore_index=True)

# Save the enriched data to processed folder
df.to_csv('../data/processed/ethiopia_fi_enriched.csv', index=False)
print("Enriched dataset saved!")
```
